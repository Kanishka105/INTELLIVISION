```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "../../../data/training/images/image2.jpg"

image = cv2.imread(image_path)

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)
```

```python
mean_3 = cv2.blur(
    gray,
    (3, 3)
)

mean_5 = cv2.blur(
    gray,
    (5, 5)
)

mean_9 = cv2.blur(
    gray,
    (9, 9)
)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(mean_3, cmap="gray")
plt.title("Mean 3x3")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(mean_5, cmap="gray")
plt.title("Mean 5x5")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(mean_9, cmap="gray")
plt.title("Mean 9x9")
plt.axis("off")

plt.show()
```

```python
print("Original:", gray.shape)
print("Mean 3x3:", mean_3.shape)
print("Mean 5x5:", mean_5.shape)
print("Mean 9x9:", mean_9.shape)
```
