```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "../../../data/training/images/image2.jpg"

image = cv2.imread(image_path)

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)
```

```python
gaussian_3 = cv2.GaussianBlur(
    gray,
    (3, 3),
    0
)

gaussian_5 = cv2.GaussianBlur(
    gray,
    (5, 5),
    0
)

gaussian_9 = cv2.GaussianBlur(
    gray,
    (9, 9),
    0
)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(gaussian_3, cmap="gray")
plt.title("Gaussian 3x3")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(gaussian_5, cmap="gray")
plt.title("Gaussian 5x5")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(gaussian_9, cmap="gray")
plt.title("Gaussian 9x9")
plt.axis("off")

plt.show()
```
