```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "../../../data/training/images/image2.jpg"

image = cv2.imread(image_path)

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Image Shape:", gray.shape)
print("Datatype:", gray.dtype)
```

```python
kernel = np.array([
    [1, 1, 1],
    [1, 1, 1],
    [1, 1, 1]
], dtype=np.float32) / 9

print("Kernel:")
print(kernel)
```

```python
convolved = cv2.filter2D(
    gray,
    -1,
    kernel
)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(convolved, cmap="gray")
plt.title("After Convolution")
plt.axis("off")

plt.show()
```

```python
row = 100
col = 100

neighborhood = gray[
    row-1:row+2,
    col-1:col+2
]

print("3x3 Neighborhood:")
print(neighborhood)

print("\nKernel:")
print(kernel)

result = np.sum(
    neighborhood * kernel
)

print("\nNew Pixel Value:", result)
```

```python
sharpen_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)
```

```python
sharpened = cv2.filter2D(
    gray,
    -1,
    sharpen_kernel
)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sharpened, cmap="gray")
plt.title("Convolution with Sharpening Kernel")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(convolved, cmap="gray")
plt.title("Mean Kernel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened, cmap="gray")
plt.title("Sharpening Kernel")
plt.axis("off")

plt.show()
```
