```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "../../../data/training/images/image2.jpg"

image = cv2.imread(image_path)

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)
```

```python
mean_filtered = cv2.blur(
    gray,
    (5, 5)
)

gaussian_filtered = cv2.GaussianBlur(
    gray,
    (5, 5),
    0
)

median_filtered = cv2.medianBlur(
    gray,
    5
)

bilateral_filtered = cv2.bilateralFilter(
    gray,
    9,
    75,
    75
)
```

```python
plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(mean_filtered, cmap="gray")
plt.title("Mean 5x5")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gaussian_filtered, cmap="gray")
plt.title("Gaussian 5x5")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(median_filtered, cmap="gray")
plt.title("Median 5x5")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(bilateral_filtered, cmap="gray")
plt.title("Bilateral")
plt.axis("off")

plt.show()
```

```python
images = {
    "Original": gray,
    "Mean": mean_filtered,
    "Gaussian": gaussian_filtered,
    "Median": median_filtered,
    "Bilateral": bilateral_filtered
}

for name, img in images.items():

    print(
        name,
        "| Mean:",
        round(float(np.mean(img)), 2),
        "| Std:",
        round(float(np.std(img)), 2),
        "| Min:",
        int(np.min(img)),
        "| Max:",
        int(np.max(img))
    )
```

```python
plt.figure(figsize=(12, 6))

plt.hist(
    gray.ravel(),
    256,
    [0, 256],
    alpha=0.5,
    label="Original"
)

plt.hist(
    mean_filtered.ravel(),
    256,
    [0, 256],
    alpha=0.5,
    label="Mean"
)

plt.hist(
    gaussian_filtered.ravel(),
    256,
    [0, 256],
    alpha=0.5,
    label="Gaussian"
)

plt.hist(
    median_filtered.ravel(),
    256,
    [0, 256],
    alpha=0.5,
    label="Median"
)

plt.hist(
    bilateral_filtered.ravel(),
    256,
    [0, 256],
    alpha=0.5,
    label="Bilateral"
)

plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.title("Filter Intensity Distribution")
plt.legend()

plt.show()
```
