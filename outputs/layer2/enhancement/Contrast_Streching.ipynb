```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "../../../data/training/images/image2.jpg"

image = cv2.imread(image_path)

gray = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)
```

```python
min_value = np.min(gray)
max_value = np.max(gray)

print("Minimum:", min_value)
print("Maximum:", max_value)
```

```python
stretched = (
    (gray - min_value) *
    255 /
    (max_value - min_value)
).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretched, cmap="gray")
plt.title("Contrast Stretching")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(
    gray.ravel(),
    256,
    [0, 256]
)
plt.title("Original Histogram")

plt.subplot(1, 2, 2)
plt.hist(
    stretched.ravel(),
    256,
    [0, 256]
)
plt.title("Stretched Histogram")

plt.show()
```
